```python
import pandas as pd
import numpy as np
from google.colab import files

print("Libraries imported successfully!")
```

```python
uploaded = files.upload()
```

```python
filename = list(uploaded.keys())[0]

if filename.lower().endswith(".csv"):
    df = pd.read_csv(filename, encoding="latin1")
elif filename.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(filename)
else:
    raise ValueError("Please upload a CSV or Excel file.")

print("Dataset loaded successfully!")
```

```python
original_rows = len(df)
print("Original row count:", original_rows)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percent.round(2)
})

print(missing_report[missing_report["Missing Values"] > 0])
if "Postal_code" in df.columns:
    df["Postal_code"] = df["Postal_code"].fillna(df["Postal_code"].median())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df[df.duplicated(keep=False)].head(10)
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicate rows removed:", before - after)
```

```python
print(df.dtypes)
```

```python
for col in df.columns:
    if col.strip().lower() in ["order date", "ship date", "order_date", "ship_date"]:
        df[col] = pd.to_datetime(df[col], errors="coerce")

print(df.dtypes)

for col in df.columns:
    if col.strip().lower() in ["order date", "ship date", "order_date", "ship_date"]:
        print(col, "Missing or invalid dates:", df[col].isna().sum())
```

```python
numeric_columns = [
    "Sales", "Quantity", "Discount", "Profit"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)
```

```python
text_columns = df.select_dtypes(include=["object"]).columns

for col in text_columns:
    df[col] = df[col].map(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Unnecessary spaces cleaned.")
```

```python
print("Remaining missing values:")
print(df.isnull().sum())
```

```python
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
print("Original row count:", original_rows)
print("Cleaned row count:", len(df))
print("Total rows removed:", original_rows - len(df))
```

```python
output_file = "Global_Superstore_Cleaned.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File name:", output_file)
```
